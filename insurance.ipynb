```python
# ==================================================
# INSURANCE DATA ANALYSIS DASHBOARD
# Error-Free Seaborn Version
# DataFrame = in1
# ==================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --------------------------------------------------
# Professional Theme
# --------------------------------------------------
sns.set_theme(style="whitegrid")

# --------------------------------------------------
# Create Dashboard
# --------------------------------------------------
fig, axes = plt.subplots(3, 2, figsize=(20, 15))

fig.suptitle(
    "Insurance Charges Analysis Dashboard",
    fontsize=24,
    fontweight="bold",
    color="#0F4C81"
)

# ==================================================
# 1. Insurance Charges Distribution
# ==================================================
sns.histplot(
    data=in1,
    x="charges",
    bins=30,
    kde=True,
    color="royalblue",
    ax=axes[0, 0]
)

axes[0, 0].set_title(
    "Distribution of Insurance Charges",
    fontsize=14,
    fontweight="bold"
)

# ==================================================
# 2. Age vs Charges
# ==================================================
sns.scatterplot(
    data=in1,
    x="age",
    y="charges",
    hue="smoker",
    palette={"yes": "#E74C3C", "no": "#2ECC71"},
    s=80,
    alpha=0.75,
    ax=axes[0, 1]
)

axes[0, 1].set_title(
    "Age vs Insurance Charges",
    fontsize=14,
    fontweight="bold"
)

# ==================================================
# 3. BMI vs Charges
# ==================================================
sns.scatterplot(
    data=in1,
    x="bmi",
    y="charges",
    hue="smoker",
    palette={"yes": "#E74C3C", "no": "#2ECC71"},
    s=80,
    alpha=0.75,
    legend=False,
    ax=axes[1, 0]
)

axes[1, 0].set_title(
    "BMI vs Insurance Charges",
    fontsize=14,
    fontweight="bold"
)

# ==================================================
# 4. Average Charges by Region
# ==================================================
region_avg = (
    in1.groupby("region", as_index=False)["charges"]
    .mean()
    .sort_values(by="charges", ascending=False)
)

sns.barplot(
    data=region_avg,
    x="region",
    y="charges",
    hue="region",       # Fix Warning
    palette="magma",
    dodge=False,
    legend=False,
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    "Average Charges by Region",
    fontsize=14,
    fontweight="bold"
)

axes[1, 1].tick_params(axis='x', rotation=20)

# ==================================================
# 5. Charges by Smoker Status
# ==================================================
sns.boxplot(
    data=in1,
    x="smoker",
    y="charges",
    hue="smoker",       # Fix Warning
    palette={
        "yes": "#E74C3C",
        "no": "#3498DB"
    },
    legend=False,
    ax=axes[2, 0]
)

axes[2, 0].set_title(
    "Charges by Smoker Status",
    fontsize=14,
    fontweight="bold"
)

# ==================================================
# 6. Correlation Heatmap
# ==================================================
corr_df = in1.copy()

corr_df["sex"] = corr_df["sex"].map({
    "male": 1,
    "female": 0
})

corr_df["smoker"] = corr_df["smoker"].map({
    "yes": 1,
    "no": 0
})

corr_df["region"] = corr_df["region"].astype("category").cat.codes

corr_matrix = corr_df.corr(numeric_only=True)

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    square=True,
    ax=axes[2, 1]
)

axes[2, 1].set_title(
    "Correlation Heatmap",
    fontsize=14,
    fontweight="bold"
)

# ==================================================
# Final Layout
# ==================================================
plt.tight_layout(rect=[0, 0, 1, 0.97])

plt.savefig(
    "Insurance_Professional_Dashboard.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ==========================================================
# ADVANCED INSURANCE ANALYSIS DASHBOARD
# DataFrame Name : in1
# No Warnings Version
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================================
# STYLE SETTINGS
# ==========================================================

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(3, 2, figsize=(20, 16))

fig.suptitle(
    "Advanced Insurance Analytics Dashboard",
    fontsize=24,
    fontweight="bold",
    color="#0B3C5D"
)

# ==========================================================
# 1. VIOLIN PLOT
# Charges Distribution by Smoker
# ==========================================================

sns.violinplot(
    data=in1,
    x="smoker",
    y="charges",
    ax=axes[0,0]
)

axes[0,0].set_title(
    "Charges Distribution by Smoker Status",
    fontsize=14,
    fontweight="bold"
)

# ==========================================================
# 2. AGE VS CHARGES REGRESSION
# ==========================================================

sns.regplot(
    data=in1,
    x="age",
    y="charges",
    scatter_kws={"alpha":0.5},
    line_kws={"color":"red"},
    ax=axes[0,1]
)

axes[0,1].set_title(
    "Age vs Charges Trend",
    fontsize=14,
    fontweight="bold"
)

# ==========================================================
# 3. BMI VS CHARGES REGRESSION
# ==========================================================

sns.regplot(
    data=in1,
    x="bmi",
    y="charges",
    scatter_kws={"alpha":0.5},
    line_kws={"color":"green"},
    ax=axes[1,0]
)

axes[1,0].set_title(
    "BMI vs Charges Trend",
    fontsize=14,
    fontweight="bold"
)

# ==========================================================
# 4. REGION DISTRIBUTION
# ==========================================================

sns.countplot(
    data=in1,
    x="region",
    ax=axes[1,1]
)

axes[1,1].set_title(
    "Customer Distribution by Region",
    fontsize=14,
    fontweight="bold"
)

axes[1,1].tick_params(axis='x', rotation=20)

# ==========================================================
# 5. CHILDREN VS AVG CHARGES
# ==========================================================

sns.pointplot(
    data=in1,
    x="children",
    y="charges",
    errorbar=None,
    ax=axes[2,0]
)

axes[2,0].set_title(
    "Average Charges by Number of Children",
    fontsize=14,
    fontweight="bold"
)

# ==========================================================
# 6. HEATMAP
# ==========================================================

corr_data = in1.copy()

corr_data["sex"] = corr_data["sex"].map({
    "male":1,
    "female":0
})

corr_data["smoker"] = corr_data["smoker"].map({
    "yes":1,
    "no":0
})

corr_data["region"] = corr_data["region"].astype("category").cat.codes

corr_matrix = corr_data.corr(numeric_only=True)

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="RdBu_r",
    fmt=".2f",
    linewidths=0.5,
    ax=axes[2,1]
)

axes[2,1].set_title(
    "Correlation Matrix",
    fontsize=14,
    fontweight="bold"
)

# ==========================================================
# FINAL LAYOUT
# ==========================================================

plt.tight_layout(rect=[0, 0, 1, 0.97])

plt.show()
```
